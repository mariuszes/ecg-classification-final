# Merge 12SL Features with PTB-XL Labels

The goal is to create one dataset that can later be used for model training.

## 1. Import libraries

```python
import pandas as pd
```

## 2. Set file paths

```python
features_path = "../data/ptb-xl-plus/12sl_features.csv"
labels_path = "../data/ptb-xl-plus/ptbxl_statements.csv"
```

## 3. Load datasets

```python
df_features = pd.read_csv(features_path)
df_labels = pd.read_csv(labels_path)

df_features.head()
```

```python
df_labels.head()
```

## 4. Check common identifier

```python
"ecg_id" in df_features.columns, "ecg_id" in df_labels.columns
```

Both datasets contain `ecg_id`, so this column can be used to connect features with labels.

## 5. Merge datasets

We merge both datasets using the `ecg_id` column.
Each row should contain ECG features and the matching diagnostic labels.

```python
df_merged = pd.merge(df_features, df_labels, on="ecg_id", how="inner")

df_merged.head()
```

## 6. Check merged dataset shape

```python
df_merged.shape
```

The merged dataset contains ECG features and diagnostic labels in one table.
This dataset can be used in the next step to prepare a target label for classification.

## 7. Save merged dataset

We save the merged dataset to a CSV file for later use.

```python
df_merged.to_csv("../data/processed/merged_12sl_ptbxl.csv", index=False)
```
