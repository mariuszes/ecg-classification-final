The file contains four columns:

- `ecg_id` - ECG record identifier, used to join labels with features.
- `scp_codes` - original diagnostic codes for the ECG record.
- `scp_codes_ext` - extended version of the diagnostic codes.
- `scp_codes_ext_snomed` - diagnostic codes mapped to SNOMED CT medical codes.

(SNOMED CT codes are standard medical codes used to describe diagnoses in the same way across different healthcare systems.)

Example interpretation:

- `ecg_id = 1` - first ECG record in the dataset.
- `NORM, 100.0` - the ECG is described as normal with certainty 100.
- `LVOLT, 100.0` - low voltage is detected with certainty 100.
- `SR, 100.0` - sinus rhythm is detected with certainty 100.

SNOMED CT (`scp_codes_ext_snomed`) - For example, `(441840, 100.0)` means that the ECG record has a specific medical label with certainty 100.
