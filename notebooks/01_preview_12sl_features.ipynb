# Preview of 12SL Features

## 1. Import libraries

```python
import pandas as pd
```

## 2. Set file path

```python
plik_12sl = "../data/ptb-xl-plus/12sl_features.csv"
```

## 3. Load the dataset

```python
df_12sl = pd.read_csv(plik_12sl)
df_12sl.head()
```

## 4. Data shape

```python
df_12sl.shape
```

The dataset contains **21,799 rows** and **783 columns**.

- Each row represents one ECG record.
- Each column represents either an identifier or a feature extracted from the ECG signal.

```python
df_12sl.columns.tolist()
```

The columns are mainly ECG morphology features.  
Their names usually contain the ECG wave or segment, the measured property, and the lead name.

Example: 

`P_Area_I` means the area of the P wave in lead I.  
`R_PeakTime_II` means the peak time of the R wave in lead II.

```python
df_12sl.info()
```

## 5. Missing values analysis

```python
df_12sl.isna().sum().sort_values(ascending=False).head(30)
```

Most columns do not contain missing values.  
The largest number of missing values appears in global P-wave features, which may mean that the P wave was not detected in some ECG records.
